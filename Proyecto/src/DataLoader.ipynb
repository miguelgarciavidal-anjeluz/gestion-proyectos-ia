```python
import random

import numpy as np
import pandas as pd

path = '../data/archive/'
accounts_path = path + 'accounts.csv'
transactions_path = path + 'transactions.csv'
df_accounts = pd.read_csv(filepath_or_buffer = accounts_path)
df_transactions = pd.read_csv(filepath_or_buffer = transactions_path)
df_transactions.drop(columns=['IS_FRAUD'], inplace=True)
df_transactions.drop(columns=['ALERT_ID'],inplace=True)


df_resultados = pd.merge(df_accounts, df_transactions, left_on='ACCOUNT_ID', right_on='SENDER_ACCOUNT_ID', how='left')
display(df_resultados.describe())
display(len(df_resultados))
```

```python
df_resultados.drop(columns=['ACCOUNT_ID'], inplace=True)
df_resultados.drop(columns=['CUSTOMER_ID'], inplace=True)
df_resultados.drop(columns=['COUNTRY'], inplace=True)
df_resultados.drop(columns=['ACCOUNT_TYPE'], inplace=True)
df_resultados.drop(columns=['TIMESTAMP'], inplace=True)
df_resultados.drop(columns=['TX_TYPE'], inplace=True)
display(len(df_resultados))
```

```python
df_resultados.dropna(inplace=True)
display(len(df_resultados))
```

```python
col1 = [random.randint(3, 7) for i in range(len(df_resultados))]
display(len(col1))
col2 = [random.randint(8, 12) for i in range(len(df_resultados))]
display(len(col2))
col3 = [random.randint(2, 5) for i in range(len(df_resultados))]
display(len(col3))
col4 = [random.randint(365, 365 * 3) for i in range(len(df_resultados))]
display(len(col4))
col5 = [random.randint(365 * 2, 365 * 8) for i in range(len(df_resultados))]
display(len(col5))

arr1 = np.array(col1)
arr2 = np.array(col2)
arr3 = np.array(col3)
arr4 = np.array(col4)
arr5 = np.array(col5)

df = pd.DataFrame(arr1, columns=['x'])
df2 = pd.DataFrame(arr2, columns=['y'])
df3 = pd.DataFrame(arr3, columns=['z'])
df4 = pd.DataFrame(arr4, columns=['w'])
df5 = pd.DataFrame(arr5, columns=['h'])

df_resultados['x'] = df
df_resultados['y'] = df2
df_resultados['z'] = df3
df_resultados['w'] = df4
df_resultados['h'] = df5

df_resultados['TX_AMOUNT'] = (df_resultados['TX_AMOUNT'] * 18.02).round(2)
df_resultados['INIT_BALANCE'] = (df_resultados['INIT_BALANCE'] * 18.02).round(2)
df_resultados['HIST_AMOUNT'] = (np.where(df_resultados['IS_FRAUD'], df_resultados['TX_AMOUNT'] * df_resultados['x'] / 10,
                                         df_resultados['TX_AMOUNT'] * df_resultados['y'] / 10)).round(2)
df_resultados['CLIENT_INCOMES'] = (df_resultados['HIST_AMOUNT'] * df_resultados['z']).round(2)
df_resultados['INCORPORATION'] = (np.where(df_resultados['IS_FRAUD'], df_resultados['w'], df_resultados['h'])).round(0)
display(len(df_resultados))
```

```python
df_resultados.drop(columns=['x'], inplace=True)
df_resultados.drop(columns=['y'], inplace=True)
df_resultados.drop(columns=['z'], inplace=True)
df_resultados.drop(columns=['w'], inplace=True)
df_resultados.drop(columns=['h'], inplace=True)
df_resultados.drop(columns=['TX_ID'], inplace=True)
df_resultados.drop(columns=['TX_BEHAVIOR_ID'], inplace=True)
df_resultados.drop(columns=['SENDER_ACCOUNT_ID'], inplace=True)
df_resultados.drop(columns=['RECEIVER_ACCOUNT_ID'], inplace=True)
df_resultados.dropna(inplace=True)
display(len(df_resultados))
```
